## Boxplot


```python
plt.figure(figsize=(3,5))

sns.boxplot(
    data=dfc,
    y= 'avg_daily_screen_time_hours'
)

plt.title('Average Daily Screen Time')
plt.ylabel('Screen Time (hours)')

plt.show()
```

Garis dalam box	Median / Q2

Batas bawah box	Q1

Batas atas box	Q3

Tinggi box	IQR = Q3 − Q1

Whisker	Rentang non-outlier

Titik di luar whisker	Potential outlier


Lower Bound = 7.5 - 1.5(2.5)
            = 3.75

Upper Bound = 10 + 1.5(2.5)
            = 13.75

## Violinplot
melihat distribusi data numerik berdasarkan kategori

```python
sns.violinplot(
    data= dfc,
    x= 'gender',
    y= 'productivity_self_rating',
    # inner= 'box', # mini boxplot dalam violin
    inner= 'quart' # garis quartile dalam violin

)

plt.title('Producitvty rating by Gender')
plt.xlabel('Gender')
plt.ylabel('Productivity Rating')

plt.show()
```

Bagian yang lebar berarti banyak data berada pada rentang tersebut.

Bagian yang sempit berarti hanya sedikit data berada di rentang tersebut. (Density)

Violinplot memberi gambaran:
"Di bagian mana data paling banyak terkonsentrasi?"

```python
plt.figure(figsize=(15,5))

sns.violinplot(
    data= dfc,
    x= 'primary_platform',
    y= 'avg_daily_screen_time_hours',
    # inner= 'box', # mini boxplot dalam violin
    inner= 'quart', # garis quartile dalam violin
    hue= 'gender',
    split= True # membagi satu violin menjadi DUA sisi berdasarkan kategori hue (hanya dua)

)

plt.title('Producitvty rating by Gender')
plt.xlabel('Gender')
plt.xticks(rotation= 75)
plt.ylabel('Productivity Rating')

plt.show()
```

## Barplot
Nilai agregat dari variabel numerik, biasanya mean (rata-rata)

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=dfc,
    x='gender',
    y='productivity_self_rating',
    errorbar= None # menghilankan bar error
)

plt.title('Average Productivity Rating by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Productivity Rating')
plt.show()
```

```python
# kalau ini pakai countplot
dfc['gender'].describe()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=dfc,
    x='gender',
    y='productivity_self_rating',
    errorbar= None, # menghilankan bar error
    estimator= 'median'
)

plt.title('Average Productivity Rating by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Productivity Rating')
plt.show()
```

Estimator ->  merubah mean jadi median

- Mean sensitif terhadap outlier.
- Median lebih robust terhadap outlier.

## Countplot VS Barplot
Countplot = berapa banyak?

Barplot = berapa nilai agregatnya?

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=dfc,
    x='primary_platform',
    hue='gender'
)

plt.title('Platform Distribution by Gender')
plt.xlabel('Primary Platform')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=dfc,
    x='primary_platform',
    y='avg_daily_screen_time_hours',
    hue='gender',
    errorbar=None
)

plt.title('Average Screen Time by Platform and Gender')
plt.xlabel('Primary Platform')
plt.ylabel('Average Screen Time (Hours)')
plt.xticks(rotation=45)
plt.show()
```

PERBEDAAN INTI

|          | `countplot`    | `barplot`           |
| -------- | -------------- | ------------------- |
| X        | Kategori       | Kategori            |
| Y        | **Count**      | **Nilai numerik**   |
| Agregasi | Jumlah data    | Mean secara default |
| `hue`    | Membagi jumlah | Membagi agregasi    |

Saat melihat chart, selalu tanyakan:

"Batang ini mewakili apa?"

Kalau countplot → jumlah observasi.

Kalau barplot → statistik agregat seperti mean/median.


## Heatmap
menampilkan nilai dalam bentuk warna, sehingga pola antar banyak variabel lebih cepat terlihat

```python
multi_cols = [
    'avg_daily_screen_time_hours',
    'avg_daily_sm_hours',
    'avg_sleep_hours',
    'avg_work_study_hours_per_day',
    'fomo_score',
    'productivity_self_rating',
    'tasks_completed_per_day'
]

corr_matrix = dfc[multi_cols].corr()
corr_matrix
```

Parameter:

corr_matrix	Data yang divisualisasika

annot=True	Menampilkan angka korelasi

fmt='.2f'	Angka dibulatkan 2 desimal

center=0	Menjadikan 0 sebagai titik tengah skala 

cmap  color map, yaitu skema warna yang digunakan untuk merepresentasikan nilai.

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    data= corr_matrix,
    annot= True,
    fmt= '.2f',
    center= 0,
    # cmap= 'coolwarm'
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## Countplot + Order
Countplot menggunakan kategori

```python
plt.figure(figsize=(15,8))

sns.countplot(
    data= dfc,
    y= 'severity_stage',
    order= dfc['severity_stage'].value_counts().index # urut dari jumlah terbesar ke terkecil
)

plt.title('Severity Stage Distribution')
plt.xlabel('Count')
plt.ylabel('Severity Stage')
plt.show()
```

## Histplot + hue

```python
gender_2 = dfc[dfc['gender'].isin(['Female', 'Male'])]
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    data=gender_2,
    x='avg_daily_screen_time_hours',
    hue='gender',
    bins=10,
    kde=True,
    fill=True, #mengisi area di bawah kurva
    alpha= 0.3 #opacity
)

plt.title('Screen Time Distribution by Gender')
plt.xlabel('Daily Screen Time (Hours)')
plt.ylabel('Count')
plt.show()
```

- Puncak kurva → area dengan density tinggi
- Kurva lebih lebar → distribusi lebih tersebar
- Kurva bergeser ke kanan → nilai cenderung lebih tinggi
- Dua kurva overlap → kedua kelompok memiliki distribusi yang mirip pada rentang tersebut
